# 7. Processos Estocásticos

## 7.1. Introdução

Os processos estocásticos constituem uma ferramenta matemática fundamental para descrever sistemas cuja evolução temporal apresenta componentes aleatórios. Diferentemente dos modelos determinísticos, nos quais o estado futuro de um sistema é completamente determinado pelas condições iniciais e pelas equações de evolução, os modelos estocásticos incorporam variáveis aleatórias para representar a incerteza associada à dinâmica do sistema.

Na Física Estatística, os processos estocásticos são utilizados para investigar fenômenos como o movimento browniano, a difusão de partículas, as flutuações térmicas e a evolução de sistemas fora do equilíbrio. Nos mercados financeiros, esses processos permitem representar matematicamente a dinâmica dos preços, dos retornos e da volatilidade dos ativos.

A utilização desses modelos possibilita estudar não apenas os valores observados de uma variável ao longo do tempo, mas também suas propriedades probabilísticas, como esperança matemática, variância, distribuição de probabilidade e dependência temporal.

Neste capítulo, serão apresentados os fundamentos matemáticos dos processos estocásticos, estabelecendo as bases para o estudo das caminhadas aleatórias, do movimento browniano e de suas aplicações à modelagem financeira.

## 7.2. Definição de Processo Estocástico

Um processo estocástico é definido como uma família de variáveis aleatórias indexadas por um parâmetro, geralmente associado ao tempo.

Matematicamente, pode ser representado por:

\[
${X_t:t\in T}$
\]

Em que:

- $(X_t)$ representa uma variável aleatória associada ao instante $(t)$;
- $(T)$ representa o conjunto de instantes considerados;
- O conjunto de possíveis valores de $(X_t)$ constitui o espaço de estados do processo.

Cada realização do processo corresponde a uma trajetória temporal específica, também denominada *realização amostral*.

É importante distinguir o processo estocástico de uma trajetória individual. O processo representa o conjunto de comportamentos probabilísticos possíveis, enquanto uma trajetória corresponde a uma sequência particular de valores observados ou simulados.

Por exemplo, considerando o preço de um ativo financeiro ao longo do tempo, podemos representar sua evolução por:

\[
${P_t:t\in T}$
\]

Cada trajetória representa uma possível evolução dos preços, enquanto o processo descreve o comportamento probabilístico associado a essas trajetórias.

## 7.3. Variáveis Aleatórias e Processos Estocásticos

Uma variável aleatória é uma função que associa resultados de um experimento aleatório a valores numéricos.

Sua distribuição de probabilidade descreve a possibilidade de ocorrência de diferentes valores.

Um processo estocástico amplia esse conceito ao considerar uma família de variáveis aleatórias relacionadas temporalmente.

Assim, enquanto uma variável aleatória descreve um fenômeno em determinado instante, um processo estocástico permite investigar sua evolução ao longo do tempo.

Essa distinção é importante porque, em sistemas dinâmicos, não basta conhecer a distribuição de uma variável isoladamente. Também é necessário investigar como seus valores se relacionam em diferentes instantes.

## 7.4. Propriedades Estatísticas dos Processos Estocásticos

A análise de processos estocásticos envolve propriedades estatísticas que permitem caracterizar seu comportamento.

### 7.4.1. Esperança Matemática

A esperança matemática representa o valor médio esperado de uma variável aleatória.

Para um processo estocástico $(X_t)$, sua esperança é definida por:

\[
$\mu_X(t)=E[X_t]$
\]

No caso de uma variável aleatória discreta:

\[
$E[X_t]=\sum_x xP(X_t=x)$
\]

Para uma variável contínua:

\[
$E[X_t]=\int_{-\infty}^{\infty}xf_{X_t}(x)\,dx$
\]

Em que $(f_{X_t}(x))$ representa a função densidade de probabilidade.

A esperança matemática permite analisar a tendência média de evolução do processo, embora não determine individualmente o comportamento de cada trajetória.

### 7.4.2. Variância

A variância quantifica a dispersão dos valores de uma variável aleatória em relação à sua esperança matemática.

É definida por:

$$
\operatorname{Var}(X_t)=E\left[(X_t-E[X_t])^2\right]
$$

Uma forma equivalente é:

$$
\operatorname{Var}(X_t)=E[X_t^2]-E[X_t]^2
$$

Em processos estocásticos, a variância permite investigar como a incerteza associada ao sistema evolui ao longo do tempo.

Esse aspecto é particularmente relevante nos mercados financeiros, nos quais a dispersão dos retornos está relacionada à volatilidade dos ativos.

### 7.4.3. Covariância

A covariância mede a relação linear entre duas variáveis aleatórias.

Para dois instantes $(t)$ e $(s)$, é definida por:

$$
\operatorname{Cov}(X_t,X_s)=
E\left[(X_t-E[X_t])(X_s-E[X_s])\right]
$$

A covariância permite investigar a dependência entre os valores de um processo em diferentes instantes.

Quando a covariância é positiva, existe uma tendência de variação conjunta no mesmo sentido. Quando é negativa, existe uma tendência de variação em sentidos opostos.

Quando é nula, não existe dependência linear, embora ainda possa existir algum tipo de dependência não linear.

### 7.4.4. Função de Autocorrelação

A função de autocorrelação mede o grau de associação linear entre os valores de um mesmo processo em diferentes instantes.

Pode ser definida por:

$$
\rho(t,s)=
\frac{\operatorname{Cov}(X_t,X_s)}
{\sqrt{\operatorname{Var}(X_t)\operatorname{Var}(X_s)}}
$$

Seu valor pertence ao intervalo:

\[
$-1\leq \rho(t,s)\leq 1 $
\]

A análise da autocorrelação é importante para identificar padrões temporais, persistência e possíveis dependências nas séries financeiras.

## 7.5. Incrementos de um Processo Estocástico

Os incrementos representam as variações de um processo entre dois instantes.

São definidos por:

\[
$\Delta X=X_{t+\Delta t}-X_t$
\]

Em um processo financeiro, por exemplo, podemos estudar a variação do preço de um ativo entre dois períodos consecutivos.

Para uma série de preços, os incrementos podem ser associados às diferenças:

\[
$\Delta P_t=P_t-P_{t-1}$
\]

Já os retornos logarítmicos são definidos por:

\[
$r_t=\ln\left(\frac{P_t}{P_{t-1}}\right)$
\]

A análise dos incrementos permite investigar características fundamentais da dinâmica de um processo, como independência, distribuição de probabilidade e comportamento da variância.

### 7.5.1. Incrementos Independentes

Um processo possui incrementos independentes quando as variações ocorridas em intervalos de tempo disjuntos são estatisticamente independentes.

Isso significa que o comportamento probabilístico de um incremento não depende dos valores assumidos por incrementos anteriores.

Formalmente, para intervalos disjuntos:

\[
$(t_1,t_2],(t_3,t_4]$
\]

com $(t_2\leq t_3)$, os incrementos correspondentes são independentes.

Essa propriedade será utilizada posteriormente no estudo da caminhada aleatória e do processo de Wiener.

É importante destacar que independência dos incrementos não significa necessariamente independência dos próprios valores do processo.

### 7.5.2. Incrementos Estacionários

Um processo possui incrementos estacionários quando a distribuição de probabilidade de um incremento depende apenas da duração do intervalo, e não do instante em que ele ocorre.

Assim:

\[
$X_{t+\Delta t}-X_t$
\]

possui a mesma distribuição para qualquer instante inicial $(t)$, desde que $\Delta t$ permaneça constante.

Essa propriedade será especialmente importante na construção de modelos contínuos de evolução temporal.

## 7.6. Estacionariedade

A estacionariedade descreve situações nas quais determinadas propriedades estatísticas de um processo permanecem invariantes ao longo do tempo.

Existem diferentes definições de estacionariedade.

### 7.6.1. Estacionariedade Estrita

Um processo é estritamente estacionário quando suas distribuições conjuntas permanecem invariantes diante de deslocamentos temporais.

Isso significa que, para qualquer conjunto de instantes $(t_1,\ldots,t_n)$ e deslocamento $(h)$:

\[
$(X_{t_1},\ldots,X_{t_n})$
\]

possui a mesma distribuição conjunta que:

\[
$(X_{t_1+h},\ldots,X_{t_n+h})$
\]

### 7.6.2. Estacionariedade Fraca

Um processo é fracamente estacionário quando apresenta:

1. Esperança matemática constante:

\[
$E[X_t]=\mu$
\]

2. Variância constante e finita:

\[
$Var(X_t)=\sigma^2$
\]

3. Covariância dependente apenas da diferença temporal:

\[
$Cov(X_t,X_s)=\gamma(t-s)$
\]

A estacionariedade é um conceito importante na análise de séries temporais financeiras, pois permite investigar se determinadas propriedades estatísticas permanecem estáveis ao longo do período analisado.

Entretanto, preços financeiros frequentemente apresentam comportamento não estacionário, enquanto os retornos podem apresentar propriedades mais próximas da estacionariedade em determinados intervalos.

## 7.7. Caminhada Aleatória (Random Walk)

A caminhada aleatória é um dos modelos mais simples de processo estocástico.

Sua característica fundamental é que a posição futura depende da posição atual acrescida de um incremento aleatório.

Pode ser representada por:

\[
$X_t=X_{t-1}+\epsilon_t$
\]

Em que:

- $(X_t)$ representa a posição do processo no instante $(t)$;
- $(X_{t-1})$ representa a posição no instante anterior;
- $(\epsilon_t)$ representa o incremento aleatório.

Considerando incrementos independentes e identicamente distribuídos, com:

\[
$E[\epsilon_t]=0$
\]

\[
$Var(\epsilon_t)=\sigma^2$
\]

podemos desenvolver as propriedades matemáticas do processo.

Expandindo recursivamente:

\[
$X_t=X_0+\sum_{i=1}^{t}\epsilon_i$
\]

### 7.7.1. Esperança Matemática da Caminhada Aleatória

Aplicando a propriedade da linearidade da esperança:

\[
$E[X_t]=E\left[X_0+\sum_{i=1}^{t}\epsilon_i\right]$
\]

Considerando $(X_0)$ constante:

\[
$E[X_t]=X_0+\sum_{i=1}^{t}E[\epsilon_i]$
\]

Como os incrementos possuem esperança nula:

\[
$\boxed{E[X_t]=X_0}$
\]

Portanto, a posição média do processo permanece constante ao longo do tempo.

### 7.7.2. Variância da Caminhada Aleatória

A variância do processo é dada por:

$$
\operatorname{Var}(X_t)=
\operatorname{Var}\left(X_0+\sum_{i=1}^{t}\epsilon_i\right)
$$

Como $(X_0)$ é constante e os incrementos são independentes:

$$
\operatorname{Var}(X_t)=
\sum_{i=1}^{t}\operatorname{Var}(\epsilon_i)
$$

Consequentemente:

$$
\boxed{\operatorname{Var}(X_t)=t\sigma^2}
$$

Esse resultado demonstra que a variância cresce linearmente com o tempo.

Portanto, mesmo que a esperança matemática permaneça constante, a dispersão das possíveis trajetórias aumenta à medida que o processo evolui.

Essa característica representa um aspecto fundamental dos processos estocásticos e será retomada no estudo do movimento browniano.

# 1.Processo de Wiener (Movimento Browniano)

## 1.1. Introdução aos Processos de Wiener

O Processo de Wiener, também conhecido como Movimento Browniano padrão, é um dos processos estocásticos mais importantes da Matemática Aplicada, da Física Estatística e das Finanças.

Ele descreve a evolução aleatória de uma variável ao longo do tempo contínuo. Seu desenvolvimento está relacionado ao estudo do movimento irregular de partículas microscópicas suspensas em um fluido, fenômeno conhecido como movimento browniano.

Diferentemente do passeio aleatório discreto, estudado anteriormente, o Processo de Wiener é definido em tempo contínuo e apresenta incrementos aleatórios cuja variância depende do intervalo temporal considerado.

Matematicamente, o processo é representado por:

$$
\{W_t,\ t\geq 0\}
$$

em que $W_t$ representa o valor do processo no instante $t$.

## 1.2. Propriedades fundamentais

O Processo de Wiener padrão apresenta cinco propriedades fundamentais:

**1. Condição inicial**

O processo começa na origem:

$$
W_0=0
$$

**2. Incrementos independentes**

Os incrementos em intervalos de tempo não sobrepostos são estatisticamente independentes.

Isso significa que a variação do processo em um intervalo não depende dos incrementos ocorridos em intervalos anteriores.

**3. Incrementos estacionários**

A distribuição dos incrementos depende apenas da duração do intervalo, e não do instante em que ele começa.

Assim, para $t>s$:

$$
W_t-W_s\sim\mathcal{N}(0,t-s)
$$

**4. Distribuição normal dos incrementos**

Os incrementos do processo seguem uma distribuição normal, com média zero e variância igual à duração do intervalo temporal.

$$
\mathbb{E}[W_t-W_s]=0
$$

$$
\operatorname{Var}(W_t-W_s)=t-s
$$

**5. Continuidade das trajetórias**

As trajetórias do Processo de Wiener são contínuas em relação ao tempo, mas não são diferenciáveis no sentido clássico.

Isso significa que o processo não apresenta saltos, embora suas oscilações sejam extremamente irregulares.

## 1.3. Média, variância e covariância

Uma das características mais importantes do Processo de Wiener é o comportamento de sua média e variância ao longo do tempo.

**Valor esperado**

A média teórica do processo é:

$$
\mathbb{E}[W_t]=0
$$

Isso indica que não existe uma tendência média de deslocamento em uma direção específica.

**Variância**

A variância aumenta linearmente com o tempo:

$$
\operatorname{Var}(W_t)=t
$$

Portanto, quanto maior o intervalo temporal, maior será a dispersão dos valores possíveis do processo.

**Covariância**

A covariância entre dois instantes é dada por:

$$
\operatorname{Cov}(W_s,W_t)=\min(s,t)
$$

Essa propriedade mostra que os valores do processo em diferentes instantes são dependentes, mesmo que seus incrementos em intervalos não sobrepostos sejam independentes.

## 1.4. Relação com o passeio aleatório

No estudo anterior, implementamos o passeio aleatório utilizando a expressão:

$$
X_t=X_{t-1}+\epsilon_t
$$

Nesse modelo, os incrementos possuem média zero e variância constante.

Para aproximar o Processo de Wiener, precisamos ajustar a escala dos incrementos de acordo com o intervalo temporal.

A expressão utilizada é:

$$
\Delta W\sim\mathcal{N}(0,\Delta t)
$$

Consequentemente, podemos escrever:

$$
W_{t+\Delta t}=W_t+\sqrt{\Delta t}\,Z
$$

em que:

- $\Delta t$ representa o intervalo temporal;
- $Z$ é uma variável aleatória com distribuição normal padrão;
- $Z\sim\mathcal{N}(0,1)$.

A presença de $\sqrt{\Delta t}$ garante que a variância dos incrementos seja proporcional ao intervalo temporal.

Essa é uma diferença fundamental em relação ao passeio aleatório discreto anteriormente implementado.

## 1.5. Relação com a Física Estatística

O Processo de Wiener possui uma relação direta com o fenômeno de difusão.

Na Física Estatística, o movimento browniano pode ser utilizado para representar o deslocamento irregular de partículas submetidas a interações microscópicas com o meio.

Uma propriedade importante é o deslocamento quadrático médio:

$$
\mathbb{E}[(W_t-W_0)^2]=t
$$

Considerando um movimento browniano físico em uma dimensão, com coeficiente de difusão $D$, temos:

$$
\mathbb{E}[(X_t-X_0)^2]=2Dt
$$

Essa relação é conhecida como equação do deslocamento quadrático médio para a difusão unidimensional.

Ela demonstra que a dispersão espacial das partículas cresce proporcionalmente ao tempo.

## 1.6. Aplicações em Finanças

O Processo de Wiener também é utilizado na modelagem matemática de mercados financeiros.

Ele constitui um dos fundamentos do Movimento Browniano Geométrico, modelo frequentemente utilizado para representar a evolução de preços de ativos financeiros.

Nesse contexto, o processo pode ser escrito por meio de uma equação diferencial estocástica:

$$
dS_t=\mu S_t\,dt+\sigma S_t\,dW_t
$$

em que:

- $S_t$ representa o preço do ativo;
- $\mu$ representa a taxa de retorno esperada;
- $\sigma$ representa a volatilidade;
- $dt$ representa o intervalo infinitesimal de tempo;
- $dW_t$ representa o incremento do Processo de Wiener.

Esse modelo será estudado posteriormente, permitindo estabelecer uma conexão entre processos estocásticos, Física Estatística e análise de séries financeiras.

Nesta etapa, o objetivo é construir uma simulação numérica do Processo de Wiener utilizando Python e NumPy.

A implementação permitirá:

2. Construir múltiplas trajetórias do movimento browniano.

4. Comparar a variância empírica com a previsão teórica.

```python
# TODO: importar as bibliotecas necessárias e implementar a análise.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Reprodutibilidade
np.random.seed(42)

# Parâmetros
X0 = 0             # Posição inicial
sigma = 1          # Desvio padrão dos incrementos
T = 1000           # Número de passos temporais
N = 10000          # Número de trajetórias

print(f"Posição inicial: {X0}")
print(f"Desvio padrão: {sigma}")
print(f"Passos temporais: {T}")
print(f"Número de trajetórias: {N}")
```

```python
# Geração dos incrementos aleatórios
incrementos = np.random.normal(
    loc=0,
    scale=sigma,
    size=(N, T)
)

print("Dimensões da matriz:", incrementos.shape)
```

```python
# Soma acumulada dos incrementos
trajetorias = X0 + np.cumsum(incrementos, axis=1)

# Incluindo a posição inicial
trajetorias = np.column_stack([
    np.full(N, X0),
    trajetorias
])

print("Dimensões das trajetórias:", trajetorias.shape)
```

## 4. Resultados
Apresentar tabelas, gráficos e medidas relevantes, com títulos, unidades e legendas.

```python
plt.figure(figsize=(12, 6))

for i in range(10):
    plt.plot(trajetorias[i], linewidth=1, alpha=0.8)

plt.axhline(X0, color='black', linestyle='--', linewidth=1)

plt.title('Caminhada Aleatória - Random Walk')
plt.xlabel('Tempo (t)')
plt.ylabel('Posição (X_t)')
plt.grid(alpha=0.3)

plt.show()
```

```python
# Média das trajetórias em cada instante
media_empirica = np.mean(trajetorias, axis=0)

# Gráfico
plt.figure(figsize=(12, 5))

plt.plot(media_empirica, label='Média empírica')

plt.axhline(
    X0,
    color='red',
    linestyle='--',
    label='Média teórica'
)

plt.title('Esperança Matemática da Caminhada Aleatória')
plt.xlabel('Tempo (t)')
plt.ylabel('E[X_t]')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
# Variância empírica
variancia_empirica = np.var(
    trajetorias,
    axis=0,
    ddof=0
)

# Variância teórica
tempo = np.arange(T + 1)

variancia_teorica = tempo * sigma**2

# Gráfico comparativo
plt.figure(figsize=(12, 6))

plt.plot(
    tempo,
    variancia_empirica,
    label='Variância empírica'
)

plt.plot(
    tempo,
    variancia_teorica,
    '--',
    label='Variância teórica'
)

plt.title('Variância da Caminhada Aleatória')
plt.xlabel('Tempo (t)')
plt.ylabel('Var(X_t)')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
instantes = [0, 10, 50, 100, 250, 500, 750, 1000]

resultados = pd.DataFrame({
    'Tempo': instantes,
    'Media_empirica': media_empirica[instantes],
    'Media_teorica': X0,
    'Variancia_empirica': variancia_empirica[instantes],
    'Variancia_teorica': variancia_teorica[instantes]
})

resultados
```

Essa tabela permitirá verificar numericamente a proximidade entre os resultados experimentais e teóricos.

# Caminhada Aleatória (Random Walk)

O objetivo desta etapa é implementar computacionalmente uma caminhada aleatória, investigando suas propriedades estatísticas e verificando a correspondência entre os resultados empíricos e as previsões teóricas.

Serão analisadas:

- Evolução temporal das trajetórias;
- Esperança matemática;
- Variância;
- Influência do número de simulações na precisão dos resultados.

```python
# Reprodutibilidade
np.random.seed(42)

# Parâmetros da simulação
X0 = 0
sigma = 1
T = 1000
N = 10000

print(f"Posição inicial: {X0}")
print(f"Desvio padrão: {sigma}")
print(f"Passos temporais: {T}")
print(f"Número de trajetórias: {N}")
```

```python
# Geração dos incrementos aleatórios
incrementos = np.random.normal(
    loc=0,
    scale=sigma,
    size=(N, T)
)

print("Dimensões da matriz:", incrementos.shape)
```
