## 7.1. Introdução à Física Estatística

A Física Estatística é uma área da Física que estuda sistemas macroscópicos a partir do comportamento coletivo de um grande número de partículas ou unidades microscópicas. Seu objetivo é estabelecer relações entre as propriedades individuais dos componentes de um sistema e as propriedades observáveis em escala macroscópica.

Um dos principais desafios dessa área consiste em compreender como interações locais entre elementos individuais podem produzir comportamentos coletivos, como ordenamento, desordem e transições de fase.

Nesse contexto, o **modelo de Ising** constitui uma das ferramentas fundamentais da Física Estatística. Originalmente desenvolvido para estudar fenômenos de magnetismo, o modelo permite investigar como a interação entre unidades microscópicas pode produzir diferentes estados de organização em um sistema.

Além de suas aplicações na Física, o modelo de Ising também apresenta possibilidades de utilização em outras áreas, como Economia, Ciências Sociais, Biologia e Ciência de Dados, especialmente em problemas que envolvem interação entre agentes, comportamento coletivo e sistemas complexos.

## 7.2. Definição do modelo de Ising

O modelo de Ising representa um sistema composto por variáveis discretas denominadas *spins*, que podem assumir exclusivamente dois estados:

$$
s_i \in \{-1,+1\}
$$

Esses estados representam duas orientações possíveis de um momento magnético.

Para estudar o comportamento coletivo, os spins são organizados em uma rede, na qual cada elemento pode interagir com seus vizinhos.

Neste projeto, utilizaremos uma rede quadrada bidimensional, na qual cada spin possui quatro vizinhos mais próximos, considerando condições periódicas de contorno.

A configuração do sistema pode ser representada por uma matriz:

$$
S =
\begin{bmatrix}
s_{1,1} & s_{1,2} & \cdots & s_{1,L}\\
s_{2,1} & s_{2,2} & \cdots & s_{2,L}\\
\vdots & \vdots & \ddots & \vdots\\
s_{L,1} & s_{L,2} & \cdots & s_{L,L}
\end{bmatrix}
$$

Em uma rede de dimensão $L \times L$, o número total de spins é:

$$
N=L^2
$$

Para nossa simulação, adotaremos uma rede de $50 \times 50$, totalizando 2.500 spins.

## 7.3. Hamiltoniano do modelo de Ising

A energia total do sistema é determinada pelo Hamiltoniano:

$$
H=-J\sum_{\langle i,j\rangle}s_i s_j-h\sum_i s_i
$$

Essa expressão apresenta dois componentes fundamentais.

**Energia de interação entre spins:**

$$
H_{\text{int}}=-J\sum_{\langle i,j\rangle}s_i s_j
$$

O parâmetro $J$ representa a intensidade da interação entre spins vizinhos.

- Quando $J>0$, a interação favorece o alinhamento dos spins, caracterizando o comportamento ferromagnético.
- Quando $J<0$, a interação favorece orientações opostas entre spins vizinhos, caracterizando o comportamento antiferromagnético.
- Quando $J=0$, não existe interação entre os spins.

**Energia associada ao campo externo:**

$$
H_{\text{campo}}=-h\sum_i s_i
$$

O parâmetro $h$ representa a intensidade do campo magnético externo, que influencia a orientação preferencial dos spins.

Quando $h>0$, existe uma preferência energética pelos estados $+1$. Quando $h<0$, a preferência ocorre pelos estados $-1$.

Neste projeto, inicialmente utilizaremos $J=1$ e $h=0$, permitindo estudar o comportamento coletivo produzido exclusivamente pelas interações entre os spins.

## 7.4. Temperatura e distribuição de Boltzmann

A temperatura é um dos parâmetros fundamentais do modelo, pois controla a influência das flutuações térmicas sobre a organização do sistema.

Na Física Estatística, a probabilidade de uma configuração de energia $H$ em equilíbrio térmico é descrita pela distribuição de Boltzmann:

$$
P(S)=\frac{e^{-\beta H(S)}}{Z}
$$

Em que:

$$
\beta=\frac{1}{k_B T}
$$

Onde:

- $T$ representa a temperatura absoluta;
- $k_B$ é a constante de Boltzmann;
- $H(S)$ representa a energia da configuração dos spins;
- $Z$ é a função de partição.

A função de partição é definida por:

$$
Z=\sum_S e^{-\beta H(S)}
$$

Ela normaliza a distribuição de probabilidades e permite determinar diversas propriedades termodinâmicas do sistema.

O comportamento do modelo depende diretamente da competição entre a energia de interação e as flutuações térmicas.

Em temperaturas baixas, a energia de interação tende a favorecer configurações mais organizadas. Em temperaturas elevadas, as flutuações térmicas favorecem configurações mais desordenadas.

## 7.5. Magnetização

A magnetização é uma das principais grandezas utilizadas para caracterizar o comportamento coletivo dos spins.

A magnetização total é definida por:

$$
M=\sum_{i=1}^{N}s_i
$$

A magnetização média por spin é:

$$
m=\frac{1}{N}\sum_{i=1}^{N}s_i
$$

Como os spins podem assumir valores $+1$ ou $-1$, a magnetização média pertence ao intervalo:

$$
-1\leq m\leq 1
$$

Sua interpretação é:

- $m\approx +1$: predominância de spins positivos.
- $m\approx -1$: predominância de spins negativos.
- $m\approx 0$: equilíbrio entre as orientações positivas e negativas.

A magnetização permite acompanhar a evolução do ordenamento do sistema durante as simulações.

É importante destacar que magnetização próxima de zero não significa necessariamente ausência de organização local, pois podem existir regiões com spins alinhados em sentidos opostos.

## 7.6. Transição de fase

Uma das propriedades mais importantes do modelo de Ising bidimensional é a existência de uma transição de fase entre estados ordenados e desordenados.

Para o modelo quadrado com interações entre primeiros vizinhos, campo externo nulo e equilíbrio termodinâmico, a temperatura crítica é dada por:

$$
\frac{k_B T_c}{J}=
\frac{2}{\ln(1+\sqrt{2})}
$$

Portanto:

$$
T_c\approx 2{,}269\frac{J}{k_B}
$$

Adotando unidades reduzidas, com $J=1$ e $k_B=1$:

$$
T_c\approx 2{,}269
$$

Abaixo da temperatura crítica, o sistema apresenta ordenamento magnético espontâneo no limite termodinâmico. Acima dela, o sistema apresenta comportamento paramagnético, caracterizado pela ausência de magnetização espontânea.

Esse fenômeno permite investigar como alterações em um parâmetro macroscópico, a temperatura, modificam o comportamento coletivo de milhares de unidades microscópicas.

## 7.7. Método de Monte Carlo e algoritmo de Metropolis

A simulação do modelo de Ising pode ser realizada por meio do método de Monte Carlo, utilizando o algoritmo de Metropolis.

Esse procedimento permite gerar configurações do sistema de acordo com a distribuição de Boltzmann, possibilitando estudar sua evolução estatística.

O algoritmo segue os seguintes princípios:

1. Seleciona-se aleatoriamente um spin da rede.
2. Considera-se a inversão do seu estado, de $+1$ para $-1$ ou vice-versa.
3. Calcula-se a variação de energia associada à alteração:

$$
\Delta E=E_{\text{novo}}-E_{\text{anterior}}
$$

4. Se $\Delta E\leq0$, a alteração é aceita.
5. Se $\Delta E>0$, a alteração é aceita com probabilidade:

$$
P_{\text{aceitação}}=e^{-\Delta E/(k_B T)}
$$

Esse mecanismo permite que o sistema explore diferentes configurações, inclusive aquelas de maior energia, respeitando a distribuição de equilíbrio térmico.

O procedimento é repetido diversas vezes até que o sistema alcance o equilíbrio estatístico.

## 7.8. Aplicação do modelo de Ising ao mercado financeiro

Além de sua aplicação original no estudo do magnetismo, o modelo de Ising pode ser utilizado como referência para investigar sistemas complexos compostos por agentes que interagem entre si.

No contexto financeiro, podemos estabelecer uma analogia na qual cada agente representa uma variável binária associada a uma decisão de mercado:

$$
s_i=
\begin{cases}
+1 & \text{Comportamento comprador}\\
-1 & \text{Comportamento vendedor}
\end{cases}
$$

Nessa interpretação, a interação entre agentes representa a influência que as decisões individuais exercem sobre o comportamento dos demais participantes.

A magnetização média pode ser interpretada como um indicador do predomínio de determinado comportamento coletivo:

$$
m=\frac{N_{\text{compradores}}-N_{\text{vendedores}}}{N}
$$

Assim, valores positivos indicariam predominância de comportamentos compradores, enquanto valores negativos indicariam predominância de comportamentos vendedores.

O parâmetro de interação $J$ poderia representar a intensidade da influência entre agentes, enquanto o campo externo $h$ poderia representar fatores externos que influenciam suas decisões.

A temperatura, por sua vez, pode ser interpretada como um parâmetro de desordem comportamental, associado à intensidade das flutuações nas decisões dos agentes.

Essa interpretação constitui uma adaptação conceitual do modelo físico, não uma equivalência direta entre temperatura termodinâmica e volatilidade financeira. Sua utilidade está em investigar como interações locais podem produzir comportamentos coletivos semelhantes aos observados em sistemas financeiros.

```python
# Reprodutibilidade
np.random.seed(42)

# Inicialização aleatória
spins = np.random.choice([-1, 1], size=(L, L))

# Estatísticas iniciais
positivos = np.sum(spins == 1)
negativos = np.sum(spins == -1)

magnetizacao = np.sum(spins)
magnetizacao_media = magnetizacao / N

print("Dimensões:", spins.shape)
print("Total de spins:", N)
print("Spins positivos:", positivos)
print("Spins negativos:", negativos)
print("Magnetização total:", magnetizacao)
print("Magnetização média:", magnetizacao_media)

# Visualização
plt.figure(figsize=(7, 7))

plt.imshow(
    spins,
    cmap="coolwarm",
    interpolation="nearest",
    vmin=-1,
    vmax=1
)

plt.colorbar(label="Estado do spin")
plt.title("Modelo de Ising - Configuração Inicial")
plt.xlabel("Posição horizontal")
plt.ylabel("Posição vertical")

plt.show()
```

```python
def calcular_energia(spins, J, h):

    vizinhos = (
        np.roll(spins, 1, axis=0) +
        np.roll(spins, -1, axis=0) +
        np.roll(spins, 1, axis=1) +
        np.roll(spins, -1, axis=1)
    )

    energia_interacao = -J * np.sum(spins * vizinhos) / 2

    energia_campo = -h * np.sum(spins)

    energia_total = energia_interacao + energia_campo

    return energia_total
```

```python
energia_inicial = calcular_energia(spins, J, h)

print("Energia inicial:", energia_inicial)
print("Energia por spin:", energia_inicial / N)
```

```python
def metropolis(spins, J, h, T):

    L = spins.shape[0]

    for _ in range(L * L):

        i = np.random.randint(0, L)
        j = np.random.randint(0, L)

        spin = spins[i, j]

        vizinhos = (
            spins[(i + 1) % L, j] +
            spins[(i - 1) % L, j] +
            spins[i, (j + 1) % L] +
            spins[i, (j - 1) % L]
        )

        delta_E = 2 * spin * (J * vizinhos + h)

        if delta_E <= 0:
            spins[i, j] *= -1

        elif np.random.random() < np.exp(-delta_E / T):
            spins[i, j] *= -1

    return spins
```

```python
# Configuração inicial
spins = np.random.choice([-1, 1], size=(L, L))

passos = 1000

energias = []
magnetizacoes = []

for passo in range(passos):

    spins = metropolis(spins, J, h, T)

    energia = calcular_energia(spins, J, h)
    magnetizacao = np.sum(spins) / N

    energias.append(energia / N)
    magnetizacoes.append(magnetizacao)

# Visualização dos resultados
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(energias)
axes[0].set_title("Energia média por spin")
axes[0].set_xlabel("Passos de Monte Carlo")
axes[0].set_ylabel("Energia")

axes[1].plot(magnetizacoes)
axes[1].set_title("Magnetização média")
axes[1].set_xlabel("Passos de Monte Carlo")
axes[1].set_ylabel("Magnetização")

plt.tight_layout()
plt.show()
```

```python
print("Resultados após a simulação")

print("Energia final:", energias[-1])
print("Magnetização final:", magnetizacoes[-1])

print("Energia média:", np.mean(energias))
print("Magnetização média:", np.mean(magnetizacoes))

print("Magnetização absoluta final:", abs(magnetizacoes[-1]))
```

```python
def simular_temperatura(T, L=50, J=1.0, h=0.0,
                        burn_in=5000, passos=5000):

    N = L * L

    # Inicialização ordenada
    spins = np.ones((L, L), dtype=int)

    # Termalização
    for _ in range(burn_in):
        spins = metropolis(spins, J, h, T)

    energias = []
    magnetizacoes = []

    # Coleta de amostras
    for _ in range(passos):

        spins = metropolis(spins, J, h, T)

        E = calcular_energia(spins, J, h)
        M = np.sum(spins)

        energias.append(E)
        magnetizacoes.append(M)

    energias = np.array(energias)
    magnetizacoes = np.array(magnetizacoes)

    # Energia média por spin
    energia_media = np.mean(energias) / N

    # Magnetização absoluta média por spin
    m_abs = np.abs(magnetizacoes) / N
    magnetizacao_media = np.mean(m_abs)

    # Calor específico
    calor_especifico = (
        np.mean(energias**2) - np.mean(energias)**2
    ) / (N * T**2)

    # Susceptibilidade com magnetização absoluta
    susceptibilidade = (
        np.mean(magnetizacoes**2)
        - np.mean(np.abs(magnetizacoes))**2
    ) / (N * T)

    return (
        energia_media,
        magnetizacao_media,
        calor_especifico,
        susceptibilidade
    )
```

```python
temperaturas = [1.0, 1.5, 2.0, 2.269, 2.5, 3.0, 4.0]

resultados = []

for T in temperaturas:

    resultado = simular_temperatura(T)

    resultados.append(resultado)

    print(f"\nTemperatura: {T}")

    print("Energia média:", resultado[0])
    print("Magnetização absoluta:", resultado[1])
    print("Calor específico:", resultado[2])
    print("Susceptibilidade:", resultado[3])
```

```python
import pandas as pd

df_ising = pd.DataFrame(
    resultados,
    columns=[
        "Energia",
        "Magnetizacao",
        "Calor_Especifico",
        "Susceptibilidade"
    ]
)

df_ising.insert(0, "Temperatura", temperaturas)

print(df_ising.to_string(index=False))
```

## Simulando em um cenário Financeiro

```python
import numpy as np
import matplotlib.pyplot as plt

L = 30
N = L * L

J = 1.0
h = 0.0
T = 2.0

np.random.seed(42)

investidores = np.random.choice(
    [-1, 1],
    size=(L, L)
)

compradores = np.sum(investidores == 1)
vendedores = np.sum(investidores == -1)

magnetizacao = np.mean(investidores)

print("Total de investidores:", N)
print("Compradores:", compradores)
print("Vendedores:", vendedores)
print("Sentimento coletivo:", magnetizacao)
```

```python
plt.figure(figsize=(7, 7))

plt.imshow(
    investidores,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Distribuição inicial dos investidores")
plt.colorbar(
    label="Estado: -1 vendedor | +1 comprador"
)

plt.show()
```

```python
def metropolis_financeiro(investidores, J, h, T):
    L = investidores.shape[0]

    for _ in range(L * L):

        i = np.random.randint(0, L)
        j = np.random.randint(0, L)

        estado = investidores[i, j]

        vizinhos = (
            investidores[(i+1) % L, j] +
            investidores[(i-1) % L, j] +
            investidores[i, (j+1) % L] +
            investidores[i, (j-1) % L]
        )

        delta_E = 2 * estado * (J * vizinhos + h)

        if delta_E <= 0:
            investidores[i, j] *= -1

        elif np.random.random() < np.exp(-delta_E / T):
            investidores[i, j] *= -1

    return investidores
```

```python
magnetizacoes = []

for periodo in range(1000):

    investidores = metropolis_financeiro(
        investidores, J, h, T
    )

    magnetizacoes.append(
        np.mean(investidores)
    )

print("Magnetização final:", np.mean(investidores))
print("Compradores:", np.sum(investidores == 1))
print("Vendedores:", np.sum(investidores == -1))
```

## Conclusão

O estudo do Modelo de Ising permitiu investigar como interações locais entre agentes podem produzir comportamentos coletivos, estabelecendo uma conexão conceitual entre Física Estatística e Mercados Financeiros.

Inicialmente, foi implementado um sistema bidimensional de spins, utilizando uma rede quadrada com condições periódicas de contorno e o algoritmo de Metropolis para simular a evolução das configurações. A partir dessa implementação, foram analisadas a energia, a magnetização, o calor específico e a susceptibilidade magnética em diferentes temperaturas.

Na etapa de análise da transição de fase, observou-se que o aumento da temperatura provoca a redução do ordenamento dos spins. Os máximos do calor específico e da susceptibilidade foram identificados próximos à temperatura crítica teórica do modelo bidimensional de Ising, aproximadamente $(T_c = 2{,}269185)$, considerando $(J=k_B=1)$. As diferenças entre os resultados numéricos e o valor teórico estão relacionadas, entre outros fatores, ao tamanho finito da rede e às limitações estatísticas das simulações.

Posteriormente, o modelo foi adaptado conceitualmente para representar investidores como agentes binários, nos quais o estado $(+1)$ representa uma decisão compradora e o estado $(-1)$, uma decisão vendedora. Nessa interpretação, a interação entre os spins passou a representar a influência social entre investidores, enquanto a magnetização foi utilizada como indicador do sentimento coletivo do sistema.

Os experimentos demonstraram que:

- **Interação entre agentes:** para $(J>0)$, a influência entre investidores favorece o alinhamento de suas decisões, representando um mecanismo simplificado de comportamento de manada.
- **Influência da temperatura:** em 4(T=1$), a magnetização média foi aproximadamente $(0{,}9788)$, indicando forte ordenamento. Em $(T=2)$, foi aproximadamente $(0{,}8267)$. Já em $(T=4)$, ficou próxima de zero, indicando maior desordem coletiva.
- **Influência do campo externo:** para $(h=+0{,}5)$, a magnetização média foi $(0{,}4536)$, enquanto para $(h=-0{,}5)$ foi $(-0{,}4477)$. Esse resultado demonstrou como influências externas podem direcionar o comportamento coletivo dos agentes.

De maneira geral, os resultados evidenciaram que regras simples de interação entre agentes podem produzir padrões coletivos complexos, mesmo quando os indivíduos possuem apenas dois estados possíveis. Essa característica torna o Modelo de Ising uma ferramenta conceitual interessante para estudar fenômenos como formação de consenso, comportamento de manada e influência de fatores externos sobre decisões coletivas.

Entretanto, a aplicação financeira desenvolvida representa uma simplificação artificial do comportamento dos mercados. A magnetização não corresponde diretamente aos retornos ou à valorização dos ativos, e a temperatura constitui um parâmetro conceitual de desordem, não uma variável econômica observável. Portanto, os resultados não devem ser interpretados como previsões de mercado, mas como uma investigação de mecanismos de interação e organização coletiva.

Conclui-se que o estudo do Modelo de Ising complementou as etapas anteriores de processos estocásticos, distribuições estatísticas e simulações de Monte Carlo, ampliando a compreensão das relações entre Física Estatística, modelagem computacional e Ciência de Dados aplicada aos mercados financeiros.
