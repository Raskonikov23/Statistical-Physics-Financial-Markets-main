# Física Estatística Aplicada aos Mercados Financeiros

## 1. Apresentação
Este projeto investiga propriedades estatísticas de séries temporais financeiras com ferramentas de Física Estatística, Matemática Aplicada e Ciência de Dados.

## 2. Problema de pesquisa
Quais propriedades estatísticas dos mercados financeiros podem ser investigadas por meio de ferramentas da Física Estatística?

## 3. Objetivos
- Analisar retornos financeiros.
- Investigar distribuições de probabilidades.
- Estudar volatilidade e correlações.
- Quantificar incerteza por meio da entropia.
- Implementar simulações de Monte Carlo.
- Explorar modelos de interação coletiva, como o modelo de Ising.

## 4. Fundamentação interdisciplinar
Mecânica Estatística, Probabilidade e Estatística, Processos Estocásticos, Econofísica e Ciência de Dados.

## 5. Ferramentas
Python, NumPy, Pandas, SciPy, Matplotlib, Seaborn, scikit-learn e yfinance.

## 6. Metodologia
A pesquisa começa pela fundamentação matemática, passa pela coleta e análise de dados e avança para modelagem estatística e simulações.

## 7. Limitações
O projeto tem finalidade educacional e exploratória. Associação estatística não demonstra causalidade, e simulações não garantem previsões confiáveis nem constituem recomendações de investimento.

## 8. Reprodutibilidade
Registrar a origem dos dados, período analisado, parâmetros, versões das bibliotecas e decisões de limpeza.

## Plano de desenvolvimento
1. Fundamentação matemática e retornos.
2. Coleta e tratamento de dados.
3. Análise exploratória.
4. Distribuições estatísticas.
5. Volatilidade e correlações.
6. Entropia financeira.
7. Processos estocásticos.
8. Simulação de Monte Carlo.
9. Modelo de Ising como experimento exploratório.
10. Conclusões, limitações e documentação.
